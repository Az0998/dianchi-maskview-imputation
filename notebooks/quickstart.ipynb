# Dianchi Mask-View Quickstart

Load the open Dianchi water-quality panel, build a spatial adjacency, and run a tiny Mask-View corruption demo.

```python
import sys
from pathlib import Path
ROOT = Path('..').resolve()
sys.path.insert(0, str(ROOT))

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from src.data import build_panel, make_windows
from src.mask_view import MaskViewConfig, generate_artificial_mask, sample_mask_views

panel = build_panel()
print('T,N,D =', panel.values.shape, 'missing% =', 100*(1-panel.mask.mean()))
print('stations:', len(panel.stations))
X, M, starts = make_windows(panel, split='test')
print('test windows', X.shape)
```

```python
cfg = MaskViewConfig(n_views=4)
views = sample_mask_views(M[:1], cfg)
fig, axes = plt.subplots(1, 4, figsize=(12, 3))
for ax, v, title in zip(axes, views, cfg.patterns):
    # show DO channel across stations
    ax.imshow(v[0, :, :, 2].T, aspect='auto', cmap='gray_r')
    ax.set_title(title)
    ax.set_xlabel('time'); ax.set_ylabel('station')
plt.suptitle('Mask-View keep-masks (white=visible)')
plt.tight_layout()
plt.show()
```

```python
import seaborn as sns
sns.heatmap(panel.adj, xticklabels=False, yticklabels=False, cmap='viridis')
plt.title('Station adjacency (distance → Gaussian kNN)')
plt.show()
```
